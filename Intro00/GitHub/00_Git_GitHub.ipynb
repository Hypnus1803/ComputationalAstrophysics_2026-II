## ¿Qué es Git?

* Sistema de **control de versiones** distribuido
* Registra el historial de cambios de tus archivos: quién cambió qué y cuándo
* Tú decides cuándo tomar un "snapshot" de los archivos del proyecto (`commit`)
* Puedes volver a cualquier versión anterior en cualquier momento
* Repositorios locales (tu computador) y remotos (GitHub)

![github flow](Git_data_flow-checkpoint.png)

## ¿Qué es GitHub?

* Plataforma en la nube que aloja repositorios Git
* Permite compartir código, colaborar y llevar un historial público (o privado)
* La usaremos para: distribuir el material del curso y **entregar tareas y proyectos**

## Comandos básicos (I)

* `git init` — inicializa un repositorio en la carpeta actual
* `git clone <url>` — clona un repositorio remoto
* `git status` — muestra el estado de tus archivos
* `git add <archivo>` / `git add .` — agrega cambios al *staging*
* `git commit -m "mensaje"` — crea un commit 

## Comandos básicos (II)

* `git push` — sube tus commits a GitHub
* `git pull` — trae los cambios más recientes desde GitHub
* `git log` — muestra el historial de commits
* `git branch` / `git checkout -b <rama>` — trabajar con ramas
* `git diff` — muestra qué cambió exactamente

## Buenas prácticas

* Haz commits **pequeños y frecuentes** (no esperes a tener cambios enormes)
* Escribe mensajes de commit claros: qué cambiaste y por qué
* Mantén la rama `main` siempre estable y funcional
* Haz `git pull` con frecuencia para evitar conflictos
* Usa un archivo `.gitignore` para excluir archivos innecesarios
